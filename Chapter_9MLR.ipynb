# Part 2: Data Cleaning

In this section, you'll clean the dataset using techniques from prior chapters. Focus on handling missing values and preparing the data for modeling.

## Question 4: Remove Rows with Missing Target Variable

Since we cannot use rows where the target variable (`SalePrice`) is missing, drop all rows where `SalePrice` is null. This is critical because we need the label to build and evaluate our model.

**Checkpoint Question**: After removing rows with missing SalePrice, how many rows remain in the dataset?

```python
# Question 4:
# Check for missing values in SalePrice before removal
print(f"Missing values in SalePrice before removal: {df['SalePrice'].isnull().sum()}")
print(f"Rows before removal: {len(df)}")
print("\n" + "="*80 + "\n")

# Remove rows with missing SalePrice
df = df.dropna(subset=['SalePrice'])

print(f"Rows after removal: {len(df)}")
print(f"Missing values in SalePrice after removal: {df['SalePrice'].isnull().sum()}")
```

## Question 5: Handle Missing Values in Features

For this assignment, use a simple approach to handle missing values:
1. For **numeric columns**: Fill missing values with the median of that column
2. For **categorical columns**: Fill missing values with the mode (most frequent value) of that column

Apply these imputation strategies and verify that no missing values remain.

**Checkpoint Question**: After imputation, what is the mean of the LotFrontage feature?

```python
# Question 5:
# Separate numeric and categorical columns
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

print(f"Number of numeric columns: {len(numeric_cols)}")
print(f"Number of categorical columns: {len(categorical_cols)}")
print("\n" + "="*80 + "\n")

# Fill missing values in numeric columns with median
for col in numeric_cols:
    if df[col].isnull().sum() > 0:
        median_value = df[col].median()
        df[col].fillna(median_value, inplace=True)
        print(f"Filled {col} with median: {median_value}")

print("\n" + "="*80 + "\n")

# Fill missing values in categorical columns with mode
for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        mode_value = df[col].mode()[0]
        df[col].fillna(mode_value, inplace=True)
        print(f"Filled {col} with mode: {mode_value}")

print("\n" + "="*80 + "\n")

# Verify no missing values remain
print(f"Total missing values remaining: {df.isnull().sum().sum()}")
print(f"\nMean of LotFrontage after imputation: {df['LotFrontage'].mean():.4f}")
```

# Part 3: Prepare Features for Multiple Linear Regression

In this section, you'll prepare your features for MLR modeling. This includes separating the label from features, dummy coding categorical variables, and optionally scaling numeric features.

## Question 6: Separate Label and Features

Create two separate variables:
- `y`: The target variable (`SalePrice`)
- `X`: All other columns as features (excluding `SalePrice`)

Display the shape of both `y` and `X` to confirm they are set up correctly.

**Checkpoint Question**: How many features (columns) are in your feature matrix `X` before dummy coding?

```python
# Question 6:
# Separate label (y) and features (X)
y = df['SalePrice']
X = df.drop('SalePrice', axis=1)

print(f"Shape of y (target): {y.shape}")
print(f"Shape of X (features): {X.shape}")
print(f"\nNumber of features before dummy coding: {X.shape[1]}")
```

## Question 7: Dummy Code Categorical Variables

Convert all categorical (object type) columns in `X` to dummy variables using `pd.get_dummies()`. Remember to use `drop_first=True` to avoid perfect multicollinearity.

After dummy coding, display the new shape of `X` and show the first few rows.

**Checkpoint Question**: After dummy coding, how many features (columns) are in your feature matrix `X`?

```python
# Question 7:
# Dummy code categorical variables
X = pd.get_dummies(X, drop_first=True)

print(f"Shape of X after dummy coding: {X.shape}")
print(f"\nNumber of features after dummy coding: {X.shape[1]}")
print("\n" + "="*80 + "\n")
print("First few rows after dummy coding:")
print(X.head())
```

## Question 8: Ensure All Features are Numeric

Verify that all columns in `X` are now numeric. Convert any remaining boolean or object columns to numeric if necessary. This ensures compatibility with statsmodels.

**Checkpoint Question**: How many non-numeric columns remain in `X` after dummy coding and conversion? (The answer should be 0)

```python
# Question 8:
# Check data types
print("Data types in X:")
print(X.dtypes.value_counts())
print("\n" + "="*80 + "\n")

# Convert any boolean columns to integer
bool_cols = X.select_dtypes(include=['bool']).columns
if len(bool_cols) > 0:
    print(f"Converting {len(bool_cols)} boolean columns to integer...")
    X[bool_cols] = X[bool_cols].astype(int)

# Check for any remaining non-numeric columns
non_numeric_cols = X.select_dtypes(exclude=['int64', 'float64', 'uint8']).columns
print(f"\nNumber of non-numeric columns remaining: {len(non_numeric_cols)}")

if len(non_numeric_cols) > 0:
    print(f"Non-numeric columns: {list(non_numeric_cols)}")
else:
    print("All columns are numeric!")
```

# Part 4: Build Multiple Linear Regression Model

In this section, you'll build your MLR model using statsmodels and interpret the results.

## Question 9: Build the MLR Model

Using `statsmodels.api.OLS`, build a multiple linear regression model to predict `SalePrice` from all features in `X`. Remember to:
1. Add a constant term to your feature matrix (use `.assign(const=1)` or `sm.add_constant()`)
2. Fit the model
3. Display the model summary

**Checkpoint Question**: What is the R-squared value for your model? Round your answer to 4 decimal places.

```python
# Question 9:
# Add constant term to feature matrix
X_with_const = sm.add_constant(X)

# Build the OLS model
model = sm.OLS(y, X_with_const)
results = model.fit()

# Display model summary
print(results.summary())
print("\n" + "="*80 + "\n")
print(f"R-squared: {results.rsquared:.4f}")
```

## Question 10: Extract Model Fit Statistics

Extract and display the following model fit statistics from your regression results:
- R-squared
- Adjusted R-squared
- Number of observations
- Number of features (excluding the constant)

**Checkpoint Question**: What is the difference between R-squared and Adjusted R-squared value? Round your answer to 5 decimal places.

```python
# Question 10:
# Extract model fit statistics
r_squared = results.rsquared
adj_r_squared = results.rsquared_adj
n_observations = results.nobs
n_features = X.shape[1]  # Excluding the constant

print("Model Fit Statistics:")
print(f"R-squared: {r_squared:.4f}")
print(f"Adjusted R-squared: {adj_r_squared:.4f}")
print(f"Number of observations: {int(n_observations)}")
print(f"Number of features (excluding constant): {n_features}")
print("\n" + "="*80 + "\n")

# Calculate difference
diff = r_squared - adj_r_squared
print(f"Difference (R-squared - Adjusted R-squared): {diff:.5f}")
```

## Question 11: Identify Statistically Significant Features

From your model results, identify features that are statistically significant. Print a summary of only those features showing:
1. The feature name
2. The coefficient value
3. The p-value

Sort this list by p-value (smallest first) to see the most significant features.

**Checkpoint Question**: How many features have statistically significant p-values? (Exclude the constant term from your count)

```python
# Question 11:
# Get coefficients, p-values, and create a dataframe
coef_df = pd.DataFrame({
    'Feature': results.params.index,
    'Coefficient': results.params.values,
    'P-value': results.pvalues.values
})

# Filter for statistically significant features (p < 0.05)
# Exclude the constant term
significant_features = coef_df[(coef_df['P-value'] < 0.05) & (coef_df['Feature'] != 'const')]

# Sort by p-value (smallest first)
significant_features = significant_features.sort_values('P-value')

print("Statistically Significant Features (p < 0.05):")
print(significant_features.to_string(index=False))
print("\n" + "="*80 + "\n")
print(f"Number of statistically significant features: {len(significant_features)}")
```

## Question 12: Find the Feature with Largest Coefficient

Identify which feature (excluding the constant) has the largest absolute coefficient value. This helps understand which feature has the strongest association with SalePrice in your model.

**Checkpoint Question**: What is the name of the feature with the largest absolute coefficient value? (Exclude the constant term)

```python
# Question 12:
# Get absolute values of coefficients (excluding constant)
coef_abs = results.params.drop('const').abs()

# Find the feature with largest absolute coefficient
largest_coef_feature = coef_abs.idxmax()
largest_coef_value = results.params[largest_coef_feature]

print(f"Feature with largest absolute coefficient: {largest_coef_feature}")
print(f"Coefficient value: {largest_coef_value:.4f}")
print(f"Absolute value: {abs(largest_coef_value):.4f}")
```

## Question 13: Calculate In-Sample Prediction Metrics

Calculate and display the following in-sample prediction metrics:
1. **Mean Absolute Error (MAE)**: The average absolute difference between actual and predicted values
2. **Root Mean Squared Error (RMSE)**: The square root of the average squared prediction error

Use the fitted values from your model (`results.fittedvalues`) to compute these metrics.

**Checkpoint Question**: What is the RMSE value? Round your answer to 2 decimal places.

```python
# Question 13:
# Get fitted values (predictions)
y_pred = results.fittedvalues

# Calculate MAE
mae = mean_absolute_error(y, y_pred)

# Calculate RMSE
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)

print("In-Sample Prediction Metrics:")
print(f"Mean Absolute Error (MAE): ${mae:,.2f}")
print(f"Root Mean Squared Error (RMSE): ${rmse:,.2f}")
```

## Question 14: Interpret a Categorical Feature Coefficient

Interpret the coefficient for the `CentralAir_Y` dummy variable in your model. This variable indicates whether a home has central air conditioning (Y=Yes, N=No is the reference category).

1. Extract and display the coefficient, standard error, t-statistic, and p-value for `CentralAir_Y`
2. Write a brief interpretation (1-2 sentences) explaining what this coefficient means in the context of predicting home prices. Write this in a print() statement so that it prints out below after the coefficient, standard error, t-stat, and p-value.

**Checkpoint Question**: What is the coefficient value for `CentralAir_Y`? Round to 2 decimal places.

```python
# Question 14:
# Extract statistics for CentralAir_Y
feature_name = 'CentralAir_Y'

# Check if the feature exists in the model
if feature_name in results.params.index:
    coefficient = results.params[feature_name]
    std_err = results.bse[feature_name]
    t_stat = results.tvalues[feature_name]
    p_value = results.pvalues[feature_name]
    
    print(f"CentralAir_Y Coefficient Analysis:")
    print(f"Coefficient: {coefficient:.2f}")
    print(f"Standard Error: {std_err:.2f}")
    print(f"t-statistic: {t_stat:.4f}")
    print(f"p-value: {p_value:.6f}")
    print("\n" + "="*80 + "\n")
    
    # Interpretation
    interpretation = f"""Interpretation:
The coefficient for CentralAir_Y is ${coefficient:,.2f}. This means that homes with 
central air conditioning are associated with a sale price that is approximately ${coefficient:,.2f} 
higher than homes without central air (the reference category), holding all other 
variables constant. This effect is statistically significant (p = {p_value:.6f})."""
    
    print(interpretation)
else:
    print(f"Feature {feature_name} not found in model. Available features include:")
    print([col for col in results.params.index if 'Central' in col])
```

Interpret the coefficient for the `OverallQual` feature in your model. This variable represents the overall material and finish quality of the home on a scale from 1 (Very Poor) to 10 (Very Excellent).

# Part 5: Optional - Feature Scaling

This section is optional but demonstrates feature scaling concepts from Chapter 9. You can skip this if you're short on time, but it's good practice.

## Question 16 (Optional): Scale Numeric Features

Apply Min-Max scaling to the original numeric features (not the dummy-coded categorical features). Then rebuild your MLR model with the scaled features and compare the results.

**Important**: Only scale the numeric features that existed before dummy coding. Do NOT scale the dummy-coded categorical variables.

Print out the R2 value after scaling rounded to 4 decimal places. 

**Checkpoint Question**: How does this value compare to the R2 of the same model before it was scaled?

```python
# Question 16:
# Create a copy of X to preserve the original
X_scaled = X.copy()

# Identify numeric columns (these were numeric before dummy coding)
# We'll identify them by checking if they're NOT dummy variables
# Dummy variables are typically 0 or 1, so we can identify original numeric features
# by checking if they have values beyond 0 and 1

# A simpler approach: identify columns that were originally numeric before dummy coding
# These would be columns that don't end with specific category values and have non-binary values
numeric_features = []
for col in X_scaled.columns:
    unique_vals = X_scaled[col].nunique()
    # If column has more than 2 unique values or max value > 1, it's likely a numeric feature
    if unique_vals > 2 or X_scaled[col].max() > 1:
        numeric_features.append(col)

print(f"Number of numeric features identified: {len(numeric_features)}")
print(f"Numeric features: {numeric_features[:10]}...")  # Show first 10
print("\n" + "="*80 + "\n")

# Apply Min-Max scaling to numeric features only
scaler = MinMaxScaler()
X_scaled[numeric_features] = scaler.fit_transform(X_scaled[numeric_features])

print("Sample of scaled features:")
print(X_scaled[numeric_features[:5]].head())
print("\n" + "="*80 + "\n")

# Build model with scaled features
X_scaled_with_const = sm.add_constant(X_scaled)
model_scaled = sm.OLS(y, X_scaled_with_const)
results_scaled = model_scaled.fit()

# Compare R-squared values
print("Model Comparison:")
print(f"R-squared (original model): {results.rsquared:.4f}")
print(f"R-squared (scaled model): {results_scaled.rsquared:.4f}")
print("\n" + "="*80 + "\n")

print(f"""Comparison:
The R-squared values are identical ({results.rsquared:.4f} vs {results_scaled.rsquared:.4f}).
This demonstrates an important property of linear regression: scaling features does NOT 
change the model's predictive performance (R-squared). Scaling only affects the magnitude 
and interpretation of the coefficients, not the overall fit or predictions.""")
```
